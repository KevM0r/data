<b>1.‑ Hacer un programa que imprima en pantalla un menu "Numeros Pares e
Impares" con la siguientes opciones:</b>
1. Ingresar numeros de inicio y de fin del rango
2. Imprimir los numeros pares del rango.
3. Imprimir los numeros Impares del rango.
4. Imprimir los numeros Primos del rango.
5. Salir


```python
import os
rango_inicio = None
rango_fin = None

def menu():
    os.system('clear')
    titulo = (" Menú: Números Pares e Impares ")
    print(titulo.center(50, '*'))
    print('\n')
    print("Selecciona una Opción")
    print("\t1. Ingresar números de inicio y de fin del rango")
    print("\t2. Imprimir los números pares del rango")
    print("\t3. Imprimir los números impares del rango")
    print("\t4. Imprimir los números Primos del rango")
    print("\t5. Salir")

def ingresar_rango():
    global rango_inicio, rango_fin
    while True:
        try:
            print("\n--- INGRESO DE RANGO ---")
            inicio = int(input("Ingresa el número de inicio del rango: "))
            fin = int(input("Ingresa el número de fin del rango: "))
            
            if inicio >= fin:
                print("\n¡ADVERTENCIA! El número de inicio debe ser menor que el número de fin.")
                input("Pulsa una tecla para reintentar...")
                continue
            
            rango_inicio = inicio
            rango_fin = fin
            print(f"\n Rango establecido: De {rango_inicio} a {rango_fin}")
            input("Pulsa una tecla para volver al menú...")
            break
        except ValueError:
            print("\n¡ERROR! Debes ingresar números enteros válidos.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\n ¡ERROR INESPERADO!: {e}")
            input("Pulsa una tecla para reintentar...")

def verificar_rango(func):
    global rango_inicio, rango_fin
    def wrapper():
        if rango_inicio is None or rango_fin is None:
            print("\n ERROR: Primero debes ingresar el rango (Opción 1).")
            input("Pulsa una tecla para volver al menú...")
            return
        func(rango_inicio, rango_fin)
    return wrapper

@verificar_rango
def imprimir_pares(inicio, fin):
    print("\n--- NÚMEROS PARES ---")
    pares = [num for num in range(inicio, fin + 1) if num % 2 == 0]
    if pares:
        print("Números pares encontrados:", pares)
    else:
        print("No se encontraron números pares en el rango.")
    input("Pulsa una tecla para volver al menú...")

@verificar_rango
def imprimir_impares(inicio, fin):
    print("\n--- NÚMEROS IMPARES ---")
    impares = [num for num in range(inicio, fin + 1) if num % 2 != 0]
    if impares:
        print("Números impares encontrados:", impares)
    else:
        print("No se encontraron números impares en el rango.")
    input("Pulsa una tecla para volver al menú...")

def es_primo(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

@verificar_rango
def imprimir_primos(inicio, fin):
    print("\n--- NÚMEROS PRIMOS ---")
    primos = [num for num in range(inicio, fin + 1) if es_primo(num)]
    if primos:
        print("Números primos encontrados:", primos)
    else:
        print("No se encontraron números primos en el rango.")
    input("Pulsa una tecla para volver al menú...")

def main():
    while True:
        menu()
        opcion = input("inserta un numero valor >> ").strip()
        
        if opcion == "1":
            ingresar_rango()
        elif opcion == "2":
            imprimir_pares()
        elif opcion == "3":
            imprimir_impares()
        elif opcion == "4":
            imprimir_primos()
        elif opcion == "5":
            print('\n\tFINALIZO EL PROGRAMA!!!!!')
            break
        else:
            print("\n¡ERROR! Opción no válida. Debes elegir entre 1, 2, 3, 4 o 5.")
            input("Pulsa una tecla para continuar...")

if __name__ == '__main__':
    main()
```

<b>2.‑ Hacer un programa que imprima en pantalla menu "Tabla de Multiplicar"
con la siguientes opciones:
</b>
1. Ingresar rango de numeros.
2. Imprimir Tabla de multiplicar solo numeros pares del rango.
3. Imprimir Tabla de multiplicar solo numeros impares del rango.
4. Ingrese numero para Imprimir su tabla de multiplicar.
5. Salir


```python
import os
rango_inicio = None
rango_fin = None

def menu():
    os.system('clear')
    titulo = (" Menú: Tablas de Multiplicar ")
    print(titulo.center(50, '*'))
    print('\n')
    print("Selecciona una Opción")
    print("\t1. Ingresar rango de números")
    print("\t2. Imprimir Tabla de multiplicar solo números pares del rango")
    print("\t3. Imprimir Tabla de multiplicar solo números impares del rango")
    print("\t4. Ingrese número para Imprimir su tabla de multiplicar")
    print("\t5. Salir")

def ingresar_rango():
    global rango_inicio, rango_fin
    while True:
        try:
            print("\n--- INGRESO DE RANGO ---")
            inicio = int(input("Ingresa el número de inicio del rango (ej. 1): "))
            fin = int(input("Ingresa el número de fin del rango (ej. 12): "))
            
            if inicio > fin:
                inicio, fin = fin, inicio
                print("\nNota: Se invirtió el orden del rango para continuar.")
            
            rango_inicio = inicio
            rango_fin = fin
            print(f"\n Rango establecido: De {rango_inicio} a {rango_fin}")
            input("Pulsa una tecla para volver al menú...")
            break
        except ValueError:
            print("\n¡ERROR! Debes ingresar números enteros válidos.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\n¡ERROR INESPERADO!: {e}")
            input("Pulsa una tecla para reintentar...")

def verificar_rango(func):
    global rango_inicio, rango_fin
    def wrapper():
        if rango_inicio is None or rango_fin is None:
            print("\n ERROR: Primero debes ingresar el rango (Opción 1).")
            input("Pulsa una tecla para volver al menú...")
            return
        func(rango_inicio, rango_fin)
    return wrapper

def imprimir_tabla(numero, limite):
    print(f"\nTabla del {numero}:")
    for i in range(1, limite + 1):
        try:
            resultado = numero * i
            print(f"{numero} x {i} = {resultado}")
        except Exception as e:
            print(f"Error al calcular: {e}")

@verificar_rango
def imprimir_pares_rango(inicio, fin):
    print("\n--- TABLAS DE MULTIPLICAR (SÓLO PARES DEL RANGO) ---")
    
    # Se define el límite de la tabla como 10, puedes ajustarlo si quieres que sea 'rango_fin'
    LIMITE_TABLA = 10 
    
    pares_encontrados = False
    for num in range(inicio, fin + 1):
        if num % 2 == 0:
            imprimir_tabla(num, LIMITE_TABLA)
            pares_encontrados = True
    
    if not pares_encontrados:
        print("No se encontraron números pares en el rango ingresado.")
        
    input("\nPulsa una tecla para volver al menú...")

@verificar_rango
def imprimir_impares_rango(inicio, fin):
    print("\n--- TABLAS DE MULTIPLICAR (SÓLO IMPARES DEL RANGO) ---")
    
    LIMITE_TABLA = 10
    
    impares_encontrados = False
    for num in range(inicio, fin + 1):
        if num % 2 != 0:
            imprimir_tabla(num, LIMITE_TABLA)
            impares_encontrados = True
            
    if not impares_encontrados:
        print("No se encontraron números impares en el rango ingresado.")
        
    input("\nPulsa una tecla para volver al menú...")

def imprimir_tabla_unica():
    print("\n--- IMPRIMIR TABLA INDIVIDUAL ---")
    while True:
        try:
            numero = int(input("Ingresa el número cuya tabla deseas imprimir: "))
            limite = int(input("Ingresa hasta qué número quieres multiplicar (ej. 10): "))
            
            imprimir_tabla(numero, limite)
            input("\nPulsa una tecla para volver al menú...")
            break
        except ValueError:
            print("\n¡ERROR! Debes ingresar números enteros válidos.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\n¡ERROR INESPERADO!: {e}")
            input("Pulsa una tecla para reintentar...")


def main():
    while True:
        menu()
        opcion = input("inserta un numero valor >> ").strip()
        
        if opcion == "1":
            ingresar_rango()
        elif opcion == "2":
            imprimir_pares_rango()
        elif opcion == "3":
            imprimir_impares_rango()
        elif opcion == "4":
            imprimir_tabla_unica()
        elif opcion == "5":
            print('\n\tFINALIZO EL PROGRAMA!!!!!')
            break
        else:
            print("\n¡ERROR! Opción no válida. Debes elegir entre 1, 2, 3, 4 o 5.")
            input("Pulsa una tecla para continuar...")

if __name__ == '__main__':
    main()
```

<b>3.‑ Hacer un programa que imprima en pantalla menu "Números Capicua"
con la siguientes opciones:
</b>
1. Ingresar un rango de numeros de hasta 4 cifras.
2. Imprimir numeros capicua de dos cifras del rango.
3. Imprimir numeros capicua de tres cifras del rango.
4. Salir


```python
import os

rango_inicio = None
rango_fin = None

def menu():
    os.system('clear')
    titulo = (" Menu: Numeros Capicua ")
    print(titulo.center(50, '*'))
    print('\n')
    print("Selecciona una Opcion")
    print("\t1. Ingresar un rango de numeros de hasta 4 cifras")
    print("\t2. Imprimir numeros capicua de dos cifras del rango")
    print("\t3. Imprimir numeros capicua de tres cifras del rango")
    print("\t4. Salir")

def ingresar_rango():
    global rango_inicio, rango_fin
    while True:
        try:
            print("\n--- INGRESO DE RANGO (HASTA 4 CIFRAS) ---")
            inicio = int(input("Ingresa el numero de inicio (max. 4 cifras): "))
            fin = int(input("Ingresa el numero de fin (max. 4 cifras): "))
            
            if not (0 <= inicio <= 9999 and 0 <= fin <= 9999):
                print("\nADVERTENCIA! Ambos numeros deben ser de hasta 4 cifras (entre 0 y 9999).")
                input("Pulsa una tecla para reintentar...")
                continue
            
            if inicio > fin:
                inicio, fin = fin, inicio
                print("\nNota: Se invirtio el orden del rango para continuar.")
            
            rango_inicio = inicio
            rango_fin = fin
            print(f"\n Rango establecido: De {rango_inicio} a {rango_fin}")
            input("Pulsa una tecla para volver al menu...")
            break
        except ValueError:
            print("\nERROR! Debes ingresar numeros enteros validos.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO!: {e}")
            input("Pulsa una tecla para reintentar...")

def es_capicua(num):
    if num < 0:
        return False
    str_num = str(num)
    return str_num == str_num[::-1]

def verificar_rango(func):
    global rango_inicio, rango_fin
    def wrapper():
        if rango_inicio is None or rango_fin is None:
            print("\n ERROR: Primero debes ingresar el rango (Opcion 1).")
            input("Pulsa una tecla para volver al menu...")
            return
        func(rango_inicio, rango_fin)
    return wrapper

@verificar_rango
def imprimir_capicua_dos_cifras(inicio, fin):
    print("\n--- NUMEROS CAPICUA DE DOS CIFRAS ---")
    capicuas = []
    
    rango_real_inicio = max(inicio, 10)
    rango_real_fin = min(fin, 99)
    
    for num in range(rango_real_inicio, rango_real_fin + 1):
        if es_capicua(num):
            capicuas.append(num)

    if capicuas:
        print(f"Capicuas de 2 cifras entre {inicio} y {fin}: {capicuas}")
    else:
        print(f"No se encontraron numeros capicua de 2 cifras en el rango de {inicio} a {fin}.")

    input("\nPulsa una tecla para volver al menu...")

@verificar_rango
def imprimir_capicua_tres_cifras(inicio, fin):
    print("\n--- NUMEROS CAPICUA DE TRES CIFRAS ---")
    capicuas = []
    
    rango_real_inicio = max(inicio, 100)
    rango_real_fin = min(fin, 999)
    
    for num in range(rango_real_inicio, rango_real_fin + 1):
        if es_capicua(num):
            capicuas.append(num)

    if capicuas:
        print(f"Capicuas de 3 cifras entre {inicio} y {fin}: {capicuas}")
    else:
        print(f"No se encontraron numeros capicua de 3 cifras en el rango de {inicio} a {fin}.")

    input("\nPulsa una tecla para volver al menu...")

def main():
    while True:
        menu()
        opcion = input("inserta un numero valor >> ").strip()
        
        if opcion == "1":
            ingresar_rango()
        elif opcion == "2":
            imprimir_capicua_dos_cifras()
        elif opcion == "3":
            imprimir_capicua_tres_cifras()
        elif opcion == "4":
            print('\n\tFINALIZO EL PROGRAMA!!!!!')
            break
        else:
            print("\nERROR! Opcion no valida. Debes elegir entre 1, 2, 3 o 4.")
            input("Pulsa una tecla para continuar...")

if __name__ == '__main__':
    main()
```

<b>4.‑ Hacer un programa que imprima en pantalla un menu de "Datos del
Postulante" que solicite ingresar las siguientes opciones:
</b>
1. Nombre (sin numeros o simbolos extraños)
2. Apellido (sin numeros o simbolos extraños)
3. DNI
4. Celular 1
5. Celular 2
6. Altura
7. Peso
8. Imprimir los datos ingresados del postulante
9. Salir

```python
import os
import re

# Diccionario global para almacenar los datos del postulante
datos_postulante = {
    "Nombre": None,
    "Apellido": None,
    "DNI": None,
    "Celular 1": None,
    "Celular 2": None,
    "Altura": None,
    "Peso": None
}

def menu():
    os.system('clear')
    titulo = (" Menú: Datos del Postulante ")
    print(titulo.center(50, '*'))
    print('\n')
    print("Selecciona una Opción")
    print("\t1. Ingresar Nombre")
    print("\t2. Ingresar Apellido")
    print("\t3. Ingresar DNI")
    print("\t4. Ingresar Celular 1")
    print("\t5. Ingresar Celular 2")
    print("\t6. Ingresar Altura (metros)")
    print("\t7. Ingresar Peso (kilogramos)")
    print("\t8. Imprimir los datos ingresados del postulante")
    print("\t9. Salir")

def validar_texto(campo):
    while True:
        try:
            valor = input(f"Ingresa {campo}: ").strip()
            # Patrón que solo permite letras, espacios y tildes/ñ
            if re.search(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ\s]', valor) or len(valor) == 0:
                raise ValueError
            
            datos_postulante[campo] = valor
            print(f"\n Campo '{campo}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        except ValueError:
            print("\nERROR! El campo no puede estar vacío y solo debe contener letras y espacios.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def validar_entero(campo, longitud=None):
    while True:
        try:
            valor = input(f"Ingresa {campo}: ").strip()
            
            # Validación de longitud (para DNI y Celulares)
            if longitud and len(valor) != longitud:
                print(f"\nERROR! El {campo} debe tener exactamente {longitud} dígitos.")
                input("Pulsa una tecla para reintentar...")
                continue
                
            numero = int(valor)
            
            if numero < 0:
                print("\nERROR! No se permiten números negativos.")
                input("Pulsa una tecla para reintentar...")
                continue
                
            datos_postulante[campo] = valor  # Almacenamos como string para mantener el formato si es DNI/Celular
            print(f"\n Campo '{campo}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        except ValueError:
            print("\nERROR! Debes ingresar un número entero válido (sin letras o símbolos).")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def validar_decimal(campo):
    while True:
        try:
            valor = input(f"Ingresa {campo}: ").strip().replace(',', '.') # Permite coma o punto decimal
            numero = float(valor)
            
            if numero <= 0:
                print("\nERROR! El valor debe ser positivo.")
                input("Pulsa una tecla para reintentar...")
                continue
                
            datos_postulante[campo] = numero
            print(f"\n Campo '{campo}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        except ValueError:
            print("\nERROR! Debes ingresar un número decimal o entero válido.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def imprimir_datos():
    os.system('clear')
    titulo = (" Datos del Postulante Ingresados ")
    print(titulo.center(50, '='))
    
    todos_ingresados = True
    for campo, valor in datos_postulante.items():
        estado = "INGRESADO"
        if valor is None:
            estado = "PENDIENTE"
            todos_ingresados = False
        print(f"{campo:<15}: {valor if valor is not None else estado}")
    
    if not todos_ingresados:
        print("\nADVERTENCIA: Faltan datos por ingresar.")
    
    print("=" * 50)
    input("Pulsa una tecla para volver al menú...")

def main():
    while True:
        menu()
        opcion = input("inserta un numero valor >> ").strip()
        
        if opcion == "1":
            validar_texto("Nombre")
        elif opcion == "2":
            validar_texto("Apellido")
        elif opcion == "3":
            validar_entero("DNI", longitud=8) # Asumimos 8 dígitos para DNI
        elif opcion == "4":
            validar_entero("Celular 1", longitud=9) # Asumimos 9 dígitos para Celular
        elif opcion == "5":
            validar_entero("Celular 2", longitud=9) # Asumimos 9 dígitos para Celular
        elif opcion == "6":
            validar_decimal("Altura")
        elif opcion == "7":
            validar_decimal("Peso")
        elif opcion == "8":
            imprimir_datos()
        elif opcion == "9":
            print('\n\tFINALIZO EL PROGRAMA!!!!!')
            break
        else:
            print("\nERROR! Opción no válida. Debes elegir entre 1 y 9.")
            input("Pulsa una tecla para continuar...")

if __name__ == '__main__':
    main()
```

<b>5.‑ Hacer un programa que imprima en pantalla un menu de "Datos del
Alumno" que solicite ingresar las siguientes opciones:</b>
1. Nombre (sin numeros o simbolos extraños)
2. Apellido (sin numeros o simbolos extraños)
3. Nota1
4. Nota2
5. Nota3
6. Nota4
7. Nota Proyecto Final
8. Imprimir promedio del alumno
9. Salir
La impresión del promedio es:
promedio = (nota1 + nota2 + nota3 + nota4 + nota proyecto final * 3) / 7
   * Si las notas: nota1 , nota2 , nota3 , nota4 ; debe tener 5 puntos extras.
   * Si promedio es mayor a 10.5, debe imprimir la palabra APROBADO, sino DESAPROBADO.
   * Si promedio es mayor a 17.5, debe imprimar adicionalmente la palabra QUINTO SUPERIOR.
   * Si promedio es mayor a 19.5, debe imprimar adicionalmente la palabra DECIMO SUPERIOR.

```python
import os
import re

# Diccionario global para almacenar los datos del alumno y sus notas
datos_alumno = {
    "Nombre": None,
    "Apellido": None,
    "Nota1": None,
    "Nota2": None,
    "Nota3": None,
    "Nota4": None,
    "Nota Proyecto Final": None
}

def menu():
    os.system('clear')
    titulo = (" Menú: Datos del Alumno ")
    print(titulo.center(50, '*'))
    print('\n')
    print("Selecciona una Opción")
    print("\t1. Ingresar Nombre")
    print("\t2. Ingresar Apellido")
    print("\t3. Ingresar Nota 1")
    print("\t4. Ingresar Nota 2")
    print("\t5. Ingresar Nota 3")
    print("\t6. Ingresar Nota 4")
    print("\t7. Ingresar Nota Proyecto Final")
    print("\t8. Imprimir promedio del alumno")
    print("\t9. Salir")

def validar_texto(campo):
    while True:
        try:
            valor = input(f"Ingresa {campo}: ").strip()
            if re.search(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ\s]', valor) or len(valor) == 0:
                raise ValueError
            
            datos_alumno[campo] = valor
            print(f"\n Campo '{campo}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        except ValueError:
            print("\nERROR! El campo no puede estar vacio y solo debe contener letras y espacios.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def validar_nota(campo):
    while True:
        try:
            valor = input(f"Ingresa {campo} (0 a 20): ").strip().replace(',', '.')
            nota = float(valor)
            
            # Las notas deben estar en un rango razonable, asumimos de 0 a 20
            if not (0 <= nota <= 20):
                print("\nERROR! La nota debe estar en el rango de 0 a 20.")
                input("Pulsa una tecla para reintentar...")
                continue
                
            datos_alumno[campo] = nota
            print(f"\n Campo '{campo}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        except ValueError:
            print("\nERROR! Debes ingresar un numero valido (entero o decimal).")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def imprimir_promedio():
    os.system('clear')
    print("--- CALCULANDO PROMEDIO ---")
    
    # 1. Verificar si todas las notas necesarias han sido ingresadas
    notas_requeridas = ["Nota1", "Nota2", "Nota3", "Nota4", "Nota Proyecto Final"]
    for nota in notas_requeridas:
        if datos_alumno[nota] is None:
            print(f"\nERROR: Falta ingresar la nota '{nota}'.")
            input("Pulsa una tecla para volver al menu...")
            return

    # 2. Asignar notas
    n1 = datos_alumno["Nota1"]
    n2 = datos_alumno["Nota2"]
    n3 = datos_alumno["Nota3"]
    n4 = datos_alumno["Nota4"]
    npf = datos_alumno["Nota Proyecto Final"]
    
    try:
        # 3. Calcular el numerador base y aplicar la condición de 5 puntos extras
        sumatoria_notas_regulares = n1 + n2 + n3 + n4
        
        # Condición de 5 puntos extra 
        # Se asume que los 5 puntos extras se aplican si las 4 notas están presentes
        # y se suman al numerador total.
        puntos_extra = 5
        
        # Numerador total según la fórmula y la condición extra
        numerador = sumatoria_notas_regulares + (npf * 3) + puntos_extra
        
        # Denominador 
        denominador = 7
        
        # Cálculo del promedio 
        promedio = numerador / denominador
        
        # 4. Imprimir resultados y condiciones
        print(f"\nNombre del Alumno: {datos_alumno['Nombre']} {datos_alumno['Apellido']}")
        print("-" * 40)
        print(f"Sumatoria de Notas Regulares: {sumatoria_notas_regulares}")
        print(f"Ponderacion Proyecto Final (x3): {npf * 3}")
        print(f"Puntos Extras aplicados: {puntos_extra}")
        print(f"Numerador Total ({numerador} / {denominador})")
        print("-" * 40)
        print(f"PROMEDIO FINAL: {promedio:.2f}")

        # 5. Aplicar condiciones de APROBADO/DESAPROBADO y MÉRITO
        
        # APROBADO/DESAPROBADO 
        if promedio > 10.5:
            resultado = "APROBADO"
        else:
            resultado = "DESAPROBADO"
        
        print(f"\nResultado Académico: {resultado}")

        # MÉRITO
        if promedio > 19.5:
            # Si es > 19.5, es automáticamente > 17.5, se imprime DECIMO SUPERIOR 
            print("MÉRITO ADICIONAL: DECIMO SUPERIOR")
        elif promedio > 17.5:
            # Si es > 17.5 y no > 19.5, se imprime QUINTO SUPERIOR 
            print("MÉRITO ADICIONAL: QUINTO SUPERIOR")
            
    except ZeroDivisionError:
        print("\nERROR DE CALCULO: El denominador es cero. Revise la formula.")
    except Exception as e:
        print(f"\nERROR INESPERADO durante el calculo: {e}")

    input("\nPulsa una tecla para volver al menu...")

def main():
    while True:
        menu()
        opcion = input("inserta un numero valor >> ").strip()
        
        if opcion == "1":
            validar_texto("Nombre")
        elif opcion == "2":
            validar_texto("Apellido")
        elif opcion == "3":
            validar_nota("Nota1")
        elif opcion == "4":
            validar_nota("Nota2")
        elif opcion == "5":
            validar_nota("Nota3")
        elif opcion == "6":
            validar_nota("Nota4")
        elif opcion == "7":
            validar_nota("Nota Proyecto Final")
        elif opcion == "8":
            imprimir_promedio()
        elif opcion == "9":
            print('\n\tFINALIZO EL PROGRAMA!!!!!')
            break
        else:
            print("\nERROR! Opcion no valida. Debes elegir entre 1 y 9.")
            input("Pulsa una tecla para continuar...")

if __name__ == '__main__':
    main()
```

<b>6.‑ Hacer un programa que imprima en pantalla menu "Numeros Capicua"
con la siguientes opciones:
</b>
1. Imprimir Numeros Capicua de 3 cifras
2. Imprimir Numeros Capicua de 4 cifras
3. Validar Numeros Capicua de 3 cifras
4. Validar Numeros Capicua de 4 cifras
5. Salir

```python
import os
import re

def menu():
    os.system('clear')
    titulo = (" Menu: Numeros Capicua ")
    print(titulo.center(50, '*'))
    print('\n')
    print("Selecciona una Opcion")
    print("\t1. Imprimir Numeros Capicua de 3 cifras")
    print("\t2. Imprimir Numeros Capicua de 4 cifras")
    print("\t3. Validar Numeros Capicua de 3 cifras")
    print("\t4. Validar Numeros Capicua de 4 cifras")
    print("\t5. Salir")

def es_capicua(num):
    # Función que verifica si un número es capicúa
    if num < 0:
        return False
    str_num = str(num)
    return str_num == str_num[::-1]

def imprimir_capicua(cifras):
    print(f"\n--- IMPRIMIR CAPICÚAS DE {cifras} CIFRAS ---")
    
    if cifras == 3:
        inicio = 100
        fin = 999
    elif cifras == 4:
        inicio = 1000
        fin = 9999
    else:
        print("\nERROR INTERNO: Cifras no soportadas.")
        input("Pulsa una tecla para volver al menu...")
        return
    
    contador = 0
    
    print("Capicúas:")
    for num in range(inicio, fin + 1):
        if es_capicua(num):
            # Imprimir directamente sin usar listas, como lo exige el ejercicio
            print(num, end=" ")
            contador += 1
            if contador % 10 == 0: # Para mantener el formato ordenado
                print()
    
    if contador == 0:
        print("No se encontraron números capicúa en el rango especificado.")
    
    print("\n")
    input("Pulsa una tecla para volver al menu...")

def validar_numero_capicua(cifras):
    while True:
        try:
            print(f"\n--- VALIDAR CAPICÚA DE {cifras} CIFRAS ---")
            
            numero_str = input(f"Ingresa un numero de {cifras} cifras: ").strip()
            
            # 1. Validar que sea un número entero
            numero = int(numero_str)
            
            # 2. Validar el número de cifras (longitud de la cadena)
            if len(numero_str) != cifras:
                print(f"\nERROR! El numero debe tener exactamente {cifras} cifras.")
                input("Pulsa una tecla para reintentar...")
                continue
                
            # 3. Validar si es capicúa
            if es_capicua(numero):
                print(f"\nEl numero {numero} SI es capicúa.")
            else:
                print(f"\nEl numero {numero} NO es capicúa.")
            
            input("Pulsa una tecla para volver al menu...")
            break
            
        except ValueError:
            print("\nERROR! Debes ingresar un numero entero valido (sin letras o simbolos).")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def main():
    while True:
        menu()
        opcion = input("inserta un numero valor >> ").strip()
        
        if opcion == "1":
            imprimir_capicua(3)
        elif opcion == "2":
            imprimir_capicua(4)
        elif opcion == "3":
            validar_numero_capicua(3)
        elif opcion == "4":
            validar_numero_capicua(4)
        elif opcion == "5":
            print('\n\tFINALIZO EL PROGRAMA!!!!!')
            break
        else:
            print("\nERROR! Opcion no valida. Debes elegir entre 1 y 5.")
            input("Pulsa una tecla para continuar...")

if __name__ == '__main__':
    main()
```

<b>7.‑ Hacer un programa que imprima en pantalla menu "Datos del Paciente"
que solicite ingresar la siguientes opciones:
</b>
1. Nombre (sin numeros o simbolos extraños)
2. Apellido (sin numeros o simbolos extraños)
3. Edad
4. Peso
5. Sexo
6. Imprimir Dieta
7. Salir

La impresión de la dieta dependera de:
Para Mujeres:
* Si su peso es menor a 40Kilos no hay dieta
* Si su peso esta entre 40kilos y 50Kilos dieta rica en vegetales (detallar dieta)
* Si su peso esta entre 50kilos y 60Kilos dieta rica en vegetales y frutas (detallar dieta)
* Si su peso esta entre 60kilos y 80Kilos dieta rica en vegetales, frutas y semillas (detallar dieta)

* Si su peso es mayor a 80Kilos debe hacerse cirugía
Para Varones:
* Si su peso es menor a 60Kilos no hay dieta
* Si su peso esta entre 60kilos y 75Kilos dieta rica en vegetales (detallar dieta)
* Si su peso esta entre 75kilos y 95Kilos dieta rica en vegetales y frutas (detallar dieta)
* Si su peso esta entre 95kilos y 120Kilos dieta rica en vegetales, frutas y semillas (detallar dieta)
* Si su peso es mayor a 120Kilos debe hacerse cirugía

```python
import os
import re

# Diccionario global para almacenar los datos del paciente
datos_paciente = {
    "Nombre": None,
    "Apellido": None,
    "Edad": None,
    "Peso": None,
    "Sexo": None
}

def menu():
    os.system('clear')
    titulo = (" Menu: Datos del Paciente ")
    print(titulo.center(50, '*'))
    print('\n')
    print("Selecciona una Opcion")
    print("\t1. Ingresar Nombre")
    print("\t2. Ingresar Apellido")
    print("\t3. Ingresar Edad (años)")
    print("\t4. Ingresar Peso (Kilos)")
    print("\t5. Ingresar Sexo (M/V)")
    print("\t6. Imprimir Dieta")
    print("\t7. Salir")

def validar_texto(campo):
    while True:
        try:
            valor = input(f"Ingresa {campo}: ").strip()
            if re.search(r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ\s]', valor) or len(valor) == 0:
                raise ValueError
            
            datos_paciente[campo] = valor
            print(f"\n Campo '{campo}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        except ValueError:
            print("\nERROR! El campo no puede estar vacio y solo debe contener letras y espacios.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def validar_entero_positivo(campo):
    while True:
        try:
            valor = input(f"Ingresa {campo}: ").strip()
            numero = int(valor)
            
            if numero <= 0:
                print("\nERROR! El valor debe ser un numero entero positivo.")
                input("Pulsa una tecla para reintentar...")
                continue
                
            datos_paciente[campo] = numero
            print(f"\n Campo '{campo}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        except ValueError:
            print("\nERROR! Debes ingresar un numero entero valido.")
            input("Pulsa una tecla para reintentar...")
        except Exception as e:
            print(f"\nERROR INESPERADO: {e}")
            input("Pulsa una tecla para reintentar...")

def validar_peso():
    # Usamos la misma lógica de entero positivo, pero para el campo Peso
    validar_entero_positivo("Peso")

def validar_sexo():
    while True:
        valor = input("Ingresa Sexo (M para Mujer, V para Varon): ").strip().upper()
        if valor in ('M', 'V'):
            datos_paciente["Sexo"] = valor
            print(f"\n Sexo '{valor}' ingresado correctamente.")
            input("Pulsa una tecla para continuar...")
            break
        else:
            print("\nERROR! Por favor, ingresa 'M' o 'V'.")
            input("Pulsa una tecla para reintentar...")

def imprimir_dieta():
    os.system('clear')
    print("--- IMPRESIÓN DE DIETA RECOMENDADA ---")

    # 1. Verificar datos mínimos
    if datos_paciente["Peso"] is None or datos_paciente["Sexo"] is None:
        print("\nERROR: Debes ingresar el Peso (Opción 4) y el Sexo (Opción 5) del paciente primero.")
        input("Pulsa una tecla para volver al menu...")
        return
        
    peso = datos_paciente["Peso"]
    sexo = datos_paciente["Sexo"]
    
    print(f"\nPaciente: {datos_paciente['Nombre']} {datos_paciente['Apellido']}")
    print(f"Peso: {peso} Kilos | Sexo: {sexo}")
    print("-" * 40)
    
    recomendacion = "Recomendacion no definida"
    
    if sexo == 'M':
        print("EVALUACIÓN PARA MUJERES:")
        if peso < 40:
            recomendacion = "No hay dieta. Mantener peso actual."
        elif 40 <= peso < 50:
            recomendacion = "Dieta rica en vegetales. Detalle: Incluir abundantes ensaladas frescas, espinacas y brócoli al vapor. Limitar carnes rojas y procesados."
        elif 50 <= peso < 60:
            recomendacion = "Dieta rica en vegetales y frutas. Detalle: Dieta vegetal base, añadiendo 3-4 porciones de fruta diaria (manzana, pera, fresas). Evitar azúcares añadidos."
        elif 60 <= peso <= 80:
            recomendacion = "Dieta rica en vegetales, frutas y semillas. Detalle: Dieta completa incluyendo frutos secos y semillas (chía, linaza) para fibra y grasas saludables. Control estricto de carbohidratos."
        else: # peso > 80
            recomendacion = "Su peso es superior a 80 Kilos. Se recomienda evaluacion para CIRUGÍA."

    elif sexo == 'V':
        print("EVALUACIÓN PARA VARONES:")
        if peso < 60:
            recomendacion = "No hay dieta. Mantener peso actual."
        elif 60 <= peso < 75:
            recomendacion = "Dieta rica en vegetales. Detalle: Incluir sopas de verduras y guisos ligeros. Fomentar proteínas magras como pollo y pescado. Limitar frituras."
        elif 75 <= peso < 95:
            recomendacion = "Dieta rica en vegetales y frutas. Detalle: Dieta vegetal base, añadir frutas cítricas y agua. Controlar estrictamente el tamaño de las porciones de carnes."
        elif 95 <= peso <= 120:
            recomendacion = "Dieta rica en vegetales, frutas y semillas. Detalle: Dieta muy estricta, incorporando granos integrales, frutos secos y semillas. Ejercicio fisico obligatorio."
        else: # peso > 120
            recomendacion = "Su peso es superior a 120 Kilos. Se recomienda evaluacion para CIRUGÍA."
    
    print(f"\nRECOMENDACIÓN: {recomendacion}")
    print("-" * 40)
    input("Pulsa una tecla para volver al menu...")

def main():
    while True:
        menu()
        opcion = input("inserta un numero valor >> ").strip()
        
        if opcion == "1":
            validar_texto("Nombre")
        elif opcion == "2":
            validar_texto("Apellido")
        elif opcion == "3":
            validar_entero_positivo("Edad")
        elif opcion == "4":
            validar_peso()
        elif opcion == "5":
            validar_sexo()
        elif opcion == "6":
            imprimir_dieta()
        elif opcion == "7":
            print('\n\tFINALIZO EL PROGRAMA!!!!!')
            break
        else:
            print("\nERROR! Opcion no valida. Debes elegir entre 1 y 7.")
            input("Pulsa una tecla para continuar...")

if __name__ == '__main__':
    main()
```
